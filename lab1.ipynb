```python

price = 250
quantity = 5
discount_rate = 10 / 100
tax_rate = 13 / 100
total_price = price * quantity
discounted_price = total_price - (total_price * discount_rate)

final_cost = discounted_price + (discounted_price * tax_rate)

# Display results
print("Total Price:", total_price)
print("Discounted Price:", discounted_price)
print("Final Cost after 13% Tax:", final_cost)
```

Write a program to check if a string is a palindrome (reads the same forwards and backwards).


```python
# Check if a string is a palindrome

text = input("Enter a string: ")

# Optional: remove spaces and make lowercase for better checking
clean_text = text.replace(" ", "").lower()

reversed_text = ""

# Reverse using for loop
for char in clean_text:
    reversed_text = char + reversed_text

# Check palindrome
if clean_text == reversed_text:
    print("The string is a palindrome.")
else:
    print("The string is not a palindrome.")
```

Write a function that calculates the sum of all numbers from 1 to n using a loop

```python
# Function to calculate sum from 1 to n using a loop

def sum_to_n(n):
    total = 0
    for i in range(1, n + 1):
        total += i
    return total

# Take input from user
num = int(input("Enter a number: "))

# Call function and display result
result = sum_to_n(num)
print("Sum from 1 to", num, "is:", result)
```

Write a function to check if a year is a leap year.

```python
def is_leap_year(year):
    return (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0)


# taking input from user
year = int(input("Enter a year: "))

# checking and printing result
if is_leap_year(year):
    print(year, "is a leap year")
else:
    print(year, "is not a leap year")
```

Write functions to find maximum and minimum in a list without using built-in max()/min()

```python
def find_max(numbers):
    max_val = numbers[0]
    for num in numbers:
        if num > max_val:
            max_val = num
    return max_val


def find_min(numbers):
    min_val = numbers[0]
    for num in numbers:
        if num < min_val:
            min_val = num
    return min_val


# taking input from user
nums = list(map(int, input("Enter numbers separated by space: ").split()))

# finding max and min
maximum = find_max(nums)
minimum = find_min(nums)

print("Maximum value:", maximum)
print("Minimum value:", minimum)
```

Write a function that counts vowels and consonants in a string

```python
def count_vowels_consonants(text):
    vowels = "aeiouAEIOU"
    vowel_count = 0
    consonant_count = 0

    for char in text:
        if char.isalpha():  # checks if it's a letter
            if char in vowels:
                vowel_count += 1
            else:
                consonant_count += 1

    return vowel_count, consonant_count


# taking input from user
string = input("Enter a string: ")

# calling function
vowels, consonants = count_vowels_consonants(string)

print("Vowels:", vowels)
print("Consonants:", consonants)
```

Write a function to reverse a string using a loop.

```python
def reverse_string(text):
    reversed_text = ""
    for char in text:
        reversed_text = char + reversed_text  # add each character in front
    return reversed_text


# taking input from user
string = input("Enter a string: ")

# calling function
result = reverse_string(string)

print("Reversed string:", result)
```

Create a number guessing game using functions and loops

```python
import random

def get_user_guess():
    return int(input("Guess a number between 1 and 100: "))

def check_guess(guess, secret):
    if guess < secret:
        print("Too low!")
        return False
    elif guess > secret:
        print("Too high!")
        return False
    else:
        print("Correct! You guessed it!")
        return True

def play_game():
    secret_number = random.randint(1, 100)
    attempts = 0
    guessed = False

    print("Welcome to the Number Guessing Game!")

    while not guessed:
        guess = get_user_guess()
        attempts += 1
        guessed = check_guess(guess, secret_number)

    print(f"You took {attempts} attempts.")

# Start the game
play_game()
```
